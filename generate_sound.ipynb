# 각속도
- 속도는 단위 시간당 이동한 거리, v = d / t
- 각속도는 단위 시간당 각이 변한 정도  ω = θ / t

# 주파수
- 시간 당 몇개의 주기가 있는가
- 시간 당 몇 번 회전하는가(360°)
- 1 주파수는 360도 각도 변화, 각속도는 시간당 각도 변화 각속도 = (360*주파수)
- ω = 2𝜋f = 2𝜋/T 

# sin함수
- y(θ) = sin (θ)
- θ = ωt , y(t) =  ωt = 2𝜋ft = 2𝜋t/T





```python
import numpy as np
from IPython.display import Audio

sample_rate = 44100 # 1초에 44100번 진동

# 0 ~ 1까지 44100번 나누고 endpoint=False 끝값을 포함하지 않는다.
# 소리 재생이 1초를 넘어가지 않기 위헤 보통 endpoint=False를 씀
time = np.linspace(0,1, sample_rate,endpoint=False)

wave = np.sin(2*np.pi*440*time)

Audio(wave, rate=sample_rate)
```
